```python
class Book:
    def __init__(self, title, author, year):
        self.title = title    
        self.author = author    
        self.year = year        


    def display_info(self):
        print(f"Tên sách: {self.title}")
        print(f"Tác giả: {self.author}")
        print(f"Năm xuất bản: {self.year}")
        print("-" * 30)



book1 = Book("Cuộc đời như 1 trò đùa", "Anh em nhà Javna", 1985)
book2 = Book("Tích cực có chừng mực", "Oliver Burkeman", 2015)

book1.display_info()
book2.display_info()
```

```python
class BankAccount:
    def __init__(self, account_number):
        self._account_number = account_number
        self._balance = 0

    def account_number(self):
        return self._account_number

    def deposit(self, amount):
        if amount > 0:
            self._balance += amount
            print(f"Nạp {amount} thành công.")
        else:
            print("Số tiền nạp không hợp lệ.")

    def withdraw(self, amount):
        if amount > self._balance:
            print("Không đủ tiền để rút.")
        elif amount <= 0:
            print("Số tiền rút không hợp lệ.")
        else:
            self._balance -= amount
            print(f"Rút {amount} thành công.")

    def get_balance(self):
        return self._balance

account = BankAccount("123456789")

print("Số tài khoản:", account.account_number)

account.deposit(2000)
account.withdraw(1600)
account.withdraw(3000)

print("Số dư hiện tại:", account.get_balance())
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_info(self):
        print(f"Tên: {self.name}")
        print(f"Tuổi: {self.age}")

class Student(Person):
    def __init__(self, name, age, student_id):
        super().__init__(name, age)
        self.student_id = student_id

    def study(self):
        print(f"Sinh viên {self.name} đang học")

student1 = Student("Ngọc Thắng", 20, "SV001")

student1.display_info()
print(f"Mã sinh viên: {student1.student_id}")
student1.study()
```

```python
class Vehicle:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def display_info(self):
        print(f"Hãng: {self.brand}")
        print(f"Mẫu: {self.model}")
        print(f"Năm: {self.year}")

class Car(Vehicle):
    def __init__(self, brand, model, year, num_doors):
        super().__init__(brand, model, year)
        self.num_doors = num_doors

    def display_info(self):
        super().display_info()
        print(f"Số cửa: {self.num_doors}")
        print("-" * 30)

class Bike(Vehicle):
    def __init__(self, brand, model, year, bike_type):
        super().__init__(brand, model, year)
        self.type = bike_type

    def display_info(self):
        super().display_info()
        print(f"Loại xe: {self.type}")
        print("-" * 30)

car1 = Car("Mazda", "CX8", 2024, 4)
bike1 = Bike("Giant", "ATX", 2022, "Mountain")

car1.display_info()
bike1.display_info()
```
