```python
# 3. Tính trung bình cộng
n = float(input("Nhập số lượng số: "))
tong = 0
for i in range(1, int(n) + 1):
    so = float(input("Nhập số: "))
    tong += so
trung_binh = tong / n
print("Trung bình cộng là:", trung_binh)
```
