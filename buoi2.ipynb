```python
# Nhập vào 1 số nguyên, kiểm tra xem số đó là số âm, dương hay bằng 0 và in kết quả tương ứng
number = int(input("Nhập vào một số nguyên: "))
if number > 0:
    print("Số đó là số dương.")
elif number < 0:
    print("Số đó là số âm.")
else:
    print("Số đó bằng 0.")
```

```python
# Viết hàm kiểm tra xem một số có phải là số nguyên tố hay không
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
number = int(input("Nhập vào một số nguyên: "))
if is_prime(number):
    print(f"{number} là số nguyên tố.")
else:
    print(f"{number} không phải là số nguyên tố.")
```

```python
# Tìm số lớn thứ 2 trong mảng
my_array = [5, 2, 9, 1, 5, 6]

max1 = my_array[0]
max2 = my_array[0]
for num in my_array:
    if num > max1:
        max2 = max1
        max1 = num
    elif num > max2 and num != max1:
        max2 = num

print("Số lớn thứ 2 trong mảng:", max2)
```

```python
# Xóa phần tử trùng lặp trong mảng, nếu không có phần trùng lặp thì hiển thị thông báo: “Không có phần từ nào trùng lặp”
my_array = [1, 2, 3, 2, 4, 1, 5]
unique_array = list(set(my_array))
if len(unique_array) == len(my_array):
    print("Không có phần từ nào trùng lặp")
else:
    print("Mảng sau khi xóa phần tử trùng lặp:", unique_array)
```

```python
# Triển khai danh sách liên kết đơn bằng class, tạo linked list
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
class LinkedList:
    def __init__(self):
        self.head = None
    def append(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        last_node = self.head
        while last_node.next:
            last_node = last_node.next
        last_node.next = new_node
    def print_list(self):
        current_node = self.head
        while current_node:
            print(current_node.data)
            current_node = current_node.next
linked_list = LinkedList()
linked_list.append(1)
linked_list.append(2)
linked_list.append(3)
linked_list.print_list()
```
